# Model testing on MRPC

```python
from datasets import load_dataset
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

from tqdm import tqdm
```

```python
def evaluate_llm(model_name, dataset, prompt_template):
    """Evaluates a decoder-only LLM on MRPC using a prompt."""
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.float16, device_map="auto")

    # Set padding token if not set
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    model.generation_config.pad_token_id = tokenizer.pad_token_id

    all_predictions = []

    model.eval()
    with torch.no_grad():
        for example in tqdm(dataset, leave=False):
            # Format the prompt
            prompt = prompt_template.format(
                sentence1=example['sentence1'],
                sentence2=example['sentence2']
            )
            # Tokenize and generate
            inputs = tokenizer(prompt, return_tensors="pt", truncation=True).to(model.device)
            outputs = model.generate(**inputs, max_new_tokens=10)
            generated_text = tokenizer.decode(outputs[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True)
            print(generated_text)

            # Parse the generated text to get a label (this logic depends on your prompt)
            if "yes" in generated_text.lower():
                predicted_label = 1
            else:
                predicted_label = 0
            all_predictions.append(predicted_label)

    return all_predictions

# Load the dataset
dataset = load_dataset('glue', 'mrpc')
# Access the split
dataset = dataset['train']

# Example prompt template for MRPC
prompt_template = """Determine if the following two sentences are semantically equivalent.
Sentence 1: {sentence1}
Sentence 2: {sentence2}
Answer (only "yes" or "no"): """
```

```python
model_names = [
    # 'Qwen/Qwen2.5-1.5b',
    # 'Qwen/Qwen2.5-3b',
    # 'Qwen/Qwen2.5-0.5b',
    # 'Qwen/Qwen3-0.6B-Base',
    # 'microsoft/phi-1_5',
    # 'microsoft/phi-1',
    # 'microsoft/phi-2',
    # "openlm-research/open_llama_3b",
]
for model_name in model_names:
    llm_predictions = evaluate_llm(model_name, dataset, prompt_template)

    # Get the true labels from the validation set
    references = dataset['label']

    tp, tn, fp, fn = 0, 0, 0, 0
    for i in range(len(llm_predictions)):
        if llm_predictions[i] == 1:
            if references[i] == 1:
                tp += 1
            else:
                fp += 1
        else:
            if references[i] == 1:
                fn += 1
            else:
                tn += 1

    print(f'Evaluated model: {model_name}')
    print(f"True positive: {tp}")
    print(f"False positive: {fp}")
    print(f"True negative: {tn}")
    print(f"False negative: {fn}")
```
