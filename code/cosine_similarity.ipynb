```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import os

def safe_tensor_load(file_path):
    """Безопасная загрузка тензора с любым dtype"""
    try:
        # Загружаем на CPU
        tensor = torch.load(file_path, map_location='cpu')

        # Конвертируем в float32 если нужно
        if tensor.dtype in [torch.bfloat16, torch.float16]:
            tensor = tensor.to(torch.float32)

        return tensor.numpy()
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return None

def load_and_convert_tensor(file_path):
    """Загружает и конвертирует тензор в float32"""
    data = torch.load(file_path, map_location='cpu')
    if data.dtype == torch.bfloat16:
        data = data.to(torch.float32)
    return data.numpy()
```

```python
def get_plots(physics_path, math_path, stat_path, model_name):
  physics_data = []
  math_data = []
  stat_data = []

  # Physics
  for file in sorted(os.listdir(physics_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(physics_path, file))
          physics_data.append(data)

  # Math
  for file in sorted(os.listdir(math_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(math_path, file))
          math_data.append(data)

  # Statistics
  for file in sorted(os.listdir(stat_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(stat_path, file))
          stat_data.append(data)

  physics_combined = np.concatenate(physics_data, axis=0)
  math_combined = np.concatenate(math_data, axis=0)
  stat_combined = np.concatenate(stat_data, axis=0)

  X = np.concatenate([physics_combined, math_combined, stat_combined], axis=0)
  y = np.concatenate([
    np.zeros(len(physics_combined)),           # physics = 0
    np.ones(len(math_combined)),              # math = 1
    np.full(len(stat_combined), 2)            # stat = 2
  ])

  pca_3d = PCA(n_components=3)
  X_pca_3d = pca_3d.fit_transform(X)

  fig = plt.figure(figsize=(14, 10))
  ax = fig.add_subplot(111, projection='3d')


  scatter1 = ax.scatter(X_pca_3d[y == 0, 0], X_pca_3d[y == 0, 1], X_pca_3d[y == 0, 2],
                       alpha=0.8, label='Physics', color='red', s=60, marker='o')

  scatter2 = ax.scatter(X_pca_3d[y == 1, 0], X_pca_3d[y == 1, 1], X_pca_3d[y == 1, 2],
                       alpha=0.8, label='Math', color='blue', s=60, marker='^')

  scatter3 = ax.scatter(X_pca_3d[y == 2, 0], X_pca_3d[y == 2, 1], X_pca_3d[y == 2, 2],
                       alpha=0.8, label='Statistics', color='green', s=60, marker='s')

  ax.set_xlabel(f'PC1 ({pca_3d.explained_variance_ratio_[0]:.2%})')
  ax.set_ylabel(f'PC2 ({pca_3d.explained_variance_ratio_[1]:.2%})')
  ax.set_zlabel(f'PC3 ({pca_3d.explained_variance_ratio_[2]:.2%})')
  title = '3D PCA: Physics vs Math vs Statistics\n(' + model_name + ', Layer 12)'
  ax.set_title(title)
  ax.legend()
  plt.tight_layout()
  plt.show()




  physics_cluster_center = np.mean(X_pca_3d[y == 0], axis=0)
  math_cluster_center = np.mean(X_pca_3d[y == 1], axis=0)
  stat_cluster_center = np.mean(X_pca_3d[y == 2], axis=0)

  return physics_cluster_center, math_cluster_center, stat_cluster_center, pca_3d.explained_variance_ratio_, pca_3d.explained_variance_ratio_.sum()
```

```python
def get_means(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path, model_name=None):
    paths = [physics_path, math_path, stat_path, cs_path, eess_path, qbio_path]
    all_data = []

    for path in paths:
        data_list = []
        for file in sorted(os.listdir(path)):
            if file.endswith('.pt'):
                data = load_and_convert_tensor(os.path.join(path, file))
                data_list.append(data)
        combined = np.concatenate(data_list, axis=0) if data_list else np.array([]).reshape(0, -1)
        all_data.append(combined)


    centroids = []
    for data in all_data:
        if data.size == 0:
            raise ValueError(f"Нет данных для одной из тем в модели {model_name}")
        centroids.append(np.mean(data, axis=0))

    return centroids


models = ["gpt2", "qwen2.5-0.5b", "qwen2.5-1.5b", "qwen3-0.6b", "qwen3-0.6b-base",
          "gemma-2b", "gemma-3-1b-it", "bloom-560m", "bloom-1b1"]

layers = {
    "gpt2": "/layer_12/",
    "qwen2.5-0.5b": "/layer_24/",
    "qwen2.5-1.5b": "/layer_28/",
    "qwen3-0.6b": "/layer_28/",
    "qwen3-0.6b-base": "/layer_28/",
    "gemma-2b": "/layer_18/",
    "gemma-3-1b-it": "/layer_26/",
    "bloom-560m": "/layer_24/",
    "bloom-1b1": "/layer_24/"
}


topic_names = ["physics", "math", "stat", "cs", "eess", "q-bio"]
topic_labels = ["Физика", "Математика", "Статистика", "CS", "EESS", "Биология"]


all_centroids = []

for m in models:
    base_path = "hidden_states/" + m + layers[m] + "arxiv/"
    physics_path = base_path + "physics"
    math_path = base_path + "math"
    stat_path = base_path + "stat"
    cs_path = base_path + "cs"
    eess_path = base_path + "eess"
    qbio_path = base_path + "q-bio"

    centroids = get_means(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path, m)
    all_centroids.append(centroids)


def metric(x1, x2):
    return np.linalg.norm(x1 - x2)

def normalize(v):
    norm = np.linalg.norm(v)
    return v / norm if norm > 0 else v

def metric(x1, x2):
    x1_n = normalize(x1)
    x2_n = normalize(x2)
    cosine_sim = np.dot(x1_n, x2_n)
    cosine_sim = np.clip(cosine_sim, -1.0, 1.0)  # защита от численных ошибок
    return 1.0 - cosine_sim


from itertools import combinations

pair_labels = []
pair_distances = []

for i, j in combinations(range(len(topic_names)), 2):
    dists = [metric(all_centroids[m_idx][i], all_centroids[m_idx][j]) for m_idx in range(len(models))]
    pair_distances.append(dists)
    pair_labels.append(f"{topic_labels[i]} – {topic_labels[j]}")


plt.figure(figsize=(14, 8))
for dists, label in zip(pair_distances, pair_labels):
    plt.plot(models, dists, marker='o', label=label)

plt.title('Попарные L2-расстояния между центроидами тем по моделям')
plt.xlabel('Модель')
plt.ylabel('L2-расстояние')
plt.xticks(rotation=45, ha="right")
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
def get_means_for_layer(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path):
    paths = [physics_path, math_path, stat_path, cs_path, eess_path, qbio_path]
    all_data = []

    for path in paths:
        data_list = []
        for file in sorted(os.listdir(path)):
            if file.endswith('.pt'):
                data = load_and_convert_tensor(os.path.join(path, file))
                data_list.append(data)
        combined = np.concatenate(data_list, axis=0) if data_list else np.array([]).reshape(0, -1)
        all_data.append(combined)

    centroids = []
    for data in all_data:
        if data.size == 0:
            raise ValueError("Нет данных для одной из тем")
        centroids.append(np.mean(data, axis=0))

    return centroids

def normalize(v):
    norm = np.linalg.norm(v)
    return v / norm if norm > 0 else v

def cosine_distance(x1, x2):
    x1_n = normalize(x1)
    x2_n = normalize(x2)
    sim = np.dot(x1_n, x2_n)
    sim = np.clip(sim, -1.0, 1.0)
    return 1.0 - sim

# Темы
topic_names = ["physics", "math", "stat", "cs", "eess", "q-bio"]
topic_labels = ["Физика", "Математика", "Статистика", "CS", "EESS", "Биология"]

# Слои GPT-2: от 0 до 11
n_layers = 12
layer_indices = list(range(1, n_layers + 1))


# Собираем центроиды для каждого слоя
all_centroids_per_layer = []

for layer in layer_indices:
    base_path = f"hidden_states/gpt2/layer_{layer}/arxiv/"
    paths = [os.path.join(base_path, t) for t in topic_names]
    centroids = get_means_for_layer(*paths)
    all_centroids_per_layer.append(centroids)

# Считаем попарные косинусные расстояния по слоям
pair_labels = []
pair_distances_per_layer = []  # [pair_idx][layer]


for i, j in combinations(range(len(topic_names)), 2):
    pair_labels.append(f"{topic_labels[i]} – {topic_labels[j]}")
    pair_distances_per_layer.append([])

# Заполняем расстояния
for layer in layer_indices:
    centroids = all_centroids_per_layer[layer - 1]
    idx = 0
    for i, j in combinations(range(len(topic_names)), 2):
        dist = cosine_distance(centroids[i], centroids[j])
        pair_distances_per_layer[idx].append(dist)
        idx += 1

# Строим график
plt.figure(figsize=(14, 8))
for dists, label in zip(pair_distances_per_layer, pair_labels):
    plt.plot(layer_indices, dists, marker='o', label=label)

plt.title('Попарные косинусные расстояния между темами в зависимости от слоя (GPT-2)')
plt.xlabel('Номер слоя')
plt.ylabel('Косинусное расстояние (1 − similarity)')
plt.xticks(layer_indices)
plt.ylim(0, 0.06)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
def get_means_for_layer(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path):
    paths = [physics_path, math_path, stat_path, cs_path, eess_path, qbio_path]
    all_data = []

    for path in paths:
        data_list = []
        for file in sorted(os.listdir(path)):
            if file.endswith('.pt'):
                data = load_and_convert_tensor(os.path.join(path, file))
                data_list.append(data)
        combined = np.concatenate(data_list, axis=0) if data_list else np.array([]).reshape(0, -1)
        all_data.append(combined)

    centroids = []
    for data in all_data:
        if data.size == 0:
            raise ValueError("Нет данных для одной из тем")
        centroids.append(np.mean(data, axis=0))

    return centroids

def normalize(v):
    norm = np.linalg.norm(v)
    return v / norm if norm > 0 else v

def cosine_distance(x1, x2):
    x1_n = normalize(x1)
    x2_n = normalize(x2)
    sim = np.dot(x1_n, x2_n)
    sim = np.clip(sim, -1.0, 1.0)
    return 1.0 - sim

# Темы
topic_names = ["physics", "math", "stat", "cs", "eess", "q-bio"]
topic_labels = ["Физика", "Математика", "Статистика", "CS", "EESS", "Биология"]

# Слои gemma2: от 1 до 18
n_layers = 18
layer_indices = list(range(1, n_layers + 1))


# Собираем центроиды для каждого слоя
all_centroids_per_layer = []

for layer in layer_indices:
    base_path = f"hidden_states/gemma-2b/layer_{layer}/arxiv/"
    paths = [os.path.join(base_path, t) for t in topic_names]
    centroids = get_means_for_layer(*paths)
    all_centroids_per_layer.append(centroids)

# Считаем попарные косинусные расстояния по слоям
pair_labels = []
pair_distances_per_layer = []  # [pair_idx][layer]
from itertools import combinations

for i, j in combinations(range(len(topic_names)), 2):
    pair_labels.append(f"{topic_labels[i]} – {topic_labels[j]}")
    pair_distances_per_layer.append([])

# Заполняем расстояния
for layer in layer_indices:
    centroids = all_centroids_per_layer[layer - 1]
    idx = 0
    for i, j in combinations(range(len(topic_names)), 2):
        dist = cosine_distance(centroids[i], centroids[j])
        pair_distances_per_layer[idx].append(dist)
        idx += 1

# Строим график
plt.figure(figsize=(14, 8))
for dists, label in zip(pair_distances_per_layer, pair_labels):
    plt.plot(layer_indices, dists, marker='o', label=label)

plt.title('Попарные косинусные расстояния между темами в зависимости от слоя (Gemma-2)')
plt.xlabel('Номер слоя')
plt.ylabel('Косинусное расстояние (1 − similarity)')
plt.xticks(layer_indices)
plt.ylim(0, 0.5)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
def get_means(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path, model_name):
  physics_data = []
  math_data = []
  stat_data = []

  cs_data = []
  eess_data = []
  qbio_data = []

  # Physics
  for file in sorted(os.listdir(physics_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(physics_path, file))
          physics_data.append(data)

  # Math
  for file in sorted(os.listdir(math_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(math_path, file))
          math_data.append(data)

  # Statistics
  for file in sorted(os.listdir(stat_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(stat_path, file))
          stat_data.append(data)

  # Computer Science
  for file in sorted(os.listdir(cs_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(cs_path, file))
          cs_data.append(data)

  # Extended Expert System Shell
  for file in sorted(os.listdir(eess_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(eess_path, file))
          eess_data.append(data)

  # Biology
  for file in sorted(os.listdir(qbio_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(qbio_path, file))
          qbio_data.append(data)

  physics_combined = np.concatenate(physics_data, axis=0)
  math_combined = np.concatenate(math_data, axis=0)
  stat_combined = np.concatenate(stat_data, axis=0)

  cs_combined = np.concatenate(cs_data, axis=0)
  eess_combined = np.concatenate(eess_data, axis=0)
  qbio_combined = np.concatenate(qbio_data, axis=0)

  X = np.concatenate([physics_combined, math_combined, stat_combined, cs_combined, eess_combined, qbio_combined], axis=0)
  y = np.concatenate([
    np.zeros(len(physics_combined)),           # physics = 0
    np.ones(len(math_combined)),              # math = 1
    np.full(len(stat_combined), 2),            # stat = 2
    np.full(len(cs_combined), 3),
    np.full(len(eess_combined), 4),
    np.full(len(qbio_combined), 5)
  ])


  physics_cluster_center = np.mean(X[y == 0], axis=0)
  math_cluster_center = np.mean(X[y == 1], axis=0)
  stat_cluster_center = np.mean(X[y == 2], axis=0)

  return physics_cluster_center, math_cluster_center, stat_cluster_center
```
