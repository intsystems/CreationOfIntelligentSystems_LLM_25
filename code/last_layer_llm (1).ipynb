```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import os

def safe_tensor_load(file_path):
    """Tensor safe load for every dtype"""
    try:
        # cpu load
        tensor = torch.load(file_path, map_location='cpu')

        # float32 convertion if needed
        if tensor.dtype in [torch.bfloat16, torch.float16]:
            tensor = tensor.to(torch.float32)

        return tensor.numpy()
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return None

def load_and_convert_tensor(file_path):
    """Tensor load  float32"""
    data = torch.load(file_path, map_location='cpu')
    if data.dtype == torch.bfloat16:
        data = data.to(torch.float32)
    return data.numpy()
```

```python
def get_means(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path, model_name=None):
    paths = [physics_path, math_path, stat_path, cs_path, eess_path, qbio_path]
    all_data = []

    for path in paths:
        data_list = []
        for file in sorted(os.listdir(path)):
            if file.endswith('.pt'):
                data = load_and_convert_tensor(os.path.join(path, file))
                data_list.append(data)
        combined = np.concatenate(data_list, axis=0) if data_list else np.array([]).reshape(0, -1)
        all_data.append(combined)


    centroids = []
    for data in all_data:
        if data.size == 0:
            raise ValueError(f"No data for any topics for {model_name}")
        centroids.append(np.mean(data, axis=0))

    return centroids


models = ["gpt2", "qwen2.5-0.5b", "qwen2.5-1.5b", "qwen3-0.6b", "qwen3-0.6b-base",
          "gemma-2b", "gemma-3-1b-it", "bloom-560m", "bloom-1b1"]

layers = {
    "gpt2": "/layer_12/",
    "qwen2.5-0.5b": "/layer_24/",
    "qwen2.5-1.5b": "/layer_28/",
    "qwen3-0.6b": "/layer_28/",
    "qwen3-0.6b-base": "/layer_28/",
    "gemma-2b": "/layer_18/",
    "gemma-3-1b-it": "/layer_26/",
    "bloom-560m": "/layer_24/",
    "bloom-1b1": "/layer_24/"
}


topic_names = ["physics", "math", "stat", "cs", "eess", "q-bio"]
topic_labels = ["Physics", "Math", "Stat", "CS", "EESS", "Q-bio"]


all_centroids = []

for m in models:
    base_path = "hidden_states/" + m + layers[m] + "arxiv/"
    physics_path = base_path + "physics"
    math_path = base_path + "math"
    stat_path = base_path + "stat"
    cs_path = base_path + "cs"
    eess_path = base_path + "eess"
    qbio_path = base_path + "q-bio"

    centroids = get_means(physics_path, math_path, stat_path, cs_path, eess_path, qbio_path, m)
    all_centroids.append(centroids)


def metric(x1, x2):
    return np.linalg.norm(x1 - x2)

def normalize(v):
    norm = np.linalg.norm(v)
    return v / norm if norm > 0 else v

def metric(x1, x2):
    x1_n = normalize(x1)
    x2_n = normalize(x2)
    cosine_sim = np.dot(x1_n, x2_n)
    cosine_sim = np.clip(cosine_sim, -1.0, 1.0)
    return 1.0 - cosine_sim


from itertools import combinations

pair_labels = []
pair_distances = []

for i, j in combinations(range(len(topic_names)), 2):
    dists = [metric(all_centroids[m_idx][i], all_centroids[m_idx][j]) for m_idx in range(len(models))]
    pair_distances.append(dists)
    pair_labels.append(f"{topic_labels[i]} – {topic_labels[j]}")


plt.figure(figsize=(14, 8))
for dists, label in zip(pair_distances, pair_labels):
    plt.plot(models, dists, marker='o', label=label)

plt.title('pairwise distances between topics centroids by model')
plt.xlabel('model name')
plt.ylabel('cosine distance')
plt.xticks(rotation=45, ha="right")
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
def get_plots(physics_path, math_path, stat_path, model_name):
  physics_data = []
  math_data = []
  stat_data = []

  # Physics
  for file in sorted(os.listdir(physics_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(physics_path, file))
          physics_data.append(data)

  # Math
  for file in sorted(os.listdir(math_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(math_path, file))
          math_data.append(data)

  # Statistics
  for file in sorted(os.listdir(stat_path)):
      if file.endswith('.pt'):
          data = load_and_convert_tensor(os.path.join(stat_path, file))
          stat_data.append(data)

  physics_combined = np.concatenate(physics_data, axis=0)
  math_combined = np.concatenate(math_data, axis=0)
  stat_combined = np.concatenate(stat_data, axis=0)

  X = np.concatenate([physics_combined, math_combined, stat_combined], axis=0)
  y = np.concatenate([
    np.zeros(len(physics_combined)),           # physics = 0
    np.ones(len(math_combined)),              # math = 1
    np.full(len(stat_combined), 2)            # stat = 2
  ])

  pca_3d = PCA(n_components=3)
  X_pca_3d = pca_3d.fit_transform(X)

  fig = plt.figure(figsize=(14, 10))
  ax = fig.add_subplot(111, projection='3d')


  scatter1 = ax.scatter(X_pca_3d[y == 0, 0], X_pca_3d[y == 0, 1], X_pca_3d[y == 0, 2],
                       alpha=0.8, label='Physics', color='red', s=60, marker='o')

  scatter2 = ax.scatter(X_pca_3d[y == 1, 0], X_pca_3d[y == 1, 1], X_pca_3d[y == 1, 2],
                       alpha=0.8, label='Math', color='blue', s=60, marker='^')

  scatter3 = ax.scatter(X_pca_3d[y == 2, 0], X_pca_3d[y == 2, 1], X_pca_3d[y == 2, 2],
                       alpha=0.8, label='Statistics', color='green', s=60, marker='s')

  ax.set_xlabel(f'PC1 ({pca_3d.explained_variance_ratio_[0]:.2%})')
  ax.set_ylabel(f'PC2 ({pca_3d.explained_variance_ratio_[1]:.2%})')
  ax.set_zlabel(f'PC3 ({pca_3d.explained_variance_ratio_[2]:.2%})')
  title = '3D PCA: Physics vs Math vs Statistics\n(' + model_name + ', Last Layer)'
  ax.set_title(title)
  ax.legend()
  plt.tight_layout()
  plt.show()




  physics_cluster_center = np.mean(X_pca_3d[y == 0], axis=0)
  math_cluster_center = np.mean(X_pca_3d[y == 1], axis=0)
  stat_cluster_center = np.mean(X_pca_3d[y == 2], axis=0)

  return physics_cluster_center, math_cluster_center, stat_cluster_center, pca_3d.explained_variance_ratio_, pca_3d.explained_variance_ratio_.sum()
```

```python
models = ["gpt2", "qwen2.5-0.5b", "qwen2.5-1.5b", "qwen3-0.6b", "qwen3-0.6b-base", "gemma-2b", "gemma-3-1b-it", "bloom-560m", "bloom-1b1"]

layers = {
    "gpt2" : "/layer_12/",
    "qwen2.5-0.5b" : "/layer_24/",
    "qwen2.5-1.5b" : "/layer_28/",
    "qwen3-0.6b" : "/layer_28/",
    "qwen3-0.6b-base" : "/layer_28/",
    "gemma-2b" : "/layer_18/",
    "gemma-3-1b-it" : "/layer_26/",
    "bloom-560m" : "/layer_24/",
    "bloom-1b1" : "/layer_24/"
}
datas = []

for m in models:
  physics_path = "hidden_states/" + m + layers[m] + "arxiv/physics"
  math_path = "hidden_states/" + m + layers[m] + "arxiv/math"
  stat_path = "hidden_states/" + m + layers[m] + "arxiv/stat"

  datas.append(get_plots(physics_path, math_path, stat_path, m))
```

```python
def get_plots(physics_path, math_path, stat_path, model_name, ax):
    """Функция для создания графика на заданной оси"""

    physics_data = []
    math_data = []
    stat_data = []

    # Physics
    for file in sorted(os.listdir(physics_path)):
        if file.endswith('.pt'):
            data = load_and_convert_tensor(os.path.join(physics_path, file))
            physics_data.append(data)

    # Math
    for file in sorted(os.listdir(math_path)):
        if file.endswith('.pt'):
            data = load_and_convert_tensor(os.path.join(math_path, file))
            math_data.append(data)

    # Statistics
    for file in sorted(os.listdir(stat_path)):
        if file.endswith('.pt'):
            data = load_and_convert_tensor(os.path.join(stat_path, file))
            stat_data.append(data)

    # Проверка на пустые данные
    if not physics_data or not math_data or not stat_data:
        print(f"Предупреждение: нет данных для модели {model_name}")
        return None, None, None, None, None

    physics_combined = np.concatenate(physics_data, axis=0)
    math_combined = np.concatenate(math_data, axis=0)
    stat_combined = np.concatenate(stat_data, axis=0)

    X = np.concatenate([physics_combined, math_combined, stat_combined], axis=0)
    y = np.concatenate([
        np.zeros(len(physics_combined)),           # physics = 0
        np.ones(len(math_combined)),              # math = 1
        np.full(len(stat_combined), 2)            # stat = 2
    ])

    pca_3d = PCA(n_components=3)
    X_pca_3d = pca_3d.fit_transform(X)

    # Разметка на заданной оси
    scatter1 = ax.scatter(X_pca_3d[y == 0, 0], X_pca_3d[y == 0, 1], X_pca_3d[y == 0, 2],
                         alpha=0.7, label='Physics', color='red', s=30, marker='o')

    scatter2 = ax.scatter(X_pca_3d[y == 1, 0], X_pca_3d[y == 1, 1], X_pca_3d[y == 1, 2],
                         alpha=0.7, label='Math', color='blue', s=30, marker='^')

    scatter3 = ax.scatter(X_pca_3d[y == 2, 0], X_pca_3d[y == 2, 1], X_pca_3d[y == 2, 2],
                         alpha=0.7, label='Statistics', color='green', s=30, marker='s')

    # Форматирование процентов
    pc1_percent = f"{pca_3d.explained_variance_ratio_[0]:.1%}"
    pc2_percent = f"{pca_3d.explained_variance_ratio_[1]:.1%}"
    pc3_percent = f"{pca_3d.explained_variance_ratio_[2]:.1%}"

    ax.set_xlabel(f'PC1\n({pc1_percent})', labelpad=10)
    ax.set_ylabel(f'PC2\n({pc2_percent})', labelpad=10)
    ax.set_zlabel(f'PC3\n({pc3_percent})', labelpad=10)

    # Укороченное название для экономии места
    short_name = model_name.replace('qwen2.5', 'q2.5').replace('qwen3', 'q3')
    ax.set_title(f'{short_name}\nLast Layer', fontsize=12, pad=10)

    # Легенда - только для первого графика
    if model_name == models[0]:
        ax.legend(loc='upper right', bbox_to_anchor=(1.3, 0.9), fontsize=9)

    # Настройка угла обзора для лучшей видимости
    ax.view_init(elev=20, azim=45)

    # Уменьшение отступов между метками и осями
    ax.xaxis.labelpad = 5
    ax.yaxis.labelpad = 5
    ax.zaxis.labelpad = 5

    physics_cluster_center = np.mean(X_pca_3d[y == 0], axis=0)
    math_cluster_center = np.mean(X_pca_3d[y == 1], axis=0)
    stat_cluster_center = np.mean(X_pca_3d[y == 2], axis=0)

    return physics_cluster_center, math_cluster_center, stat_cluster_center, \
           pca_3d.explained_variance_ratio_, pca_3d.explained_variance_ratio_.sum()
```

```python
fig = plt.figure(figsize=(20, 15))  # Увеличил размер для 9 графиков
datas = []

# Создание подграфиков
for i, model_name in enumerate(models):
    row = i // 3
    col = i % 3

    # Создание 3D подграфика
    ax = fig.add_subplot(3, 3, i+1, projection='3d')

    # Формирование путей
    physics_path = f"hidden_states/{model_name}{layers[model_name]}arxiv/physics"
    math_path = f"hidden_states/{model_name}{layers[model_name]}arxiv/math"
    stat_path = f"hidden_states/{model_name}{layers[model_name]}arxiv/stat"

    # Проверка существования путей
    if not os.path.exists(physics_path):
        print(f"Предупреждение: путь не существует - {physics_path}")
        continue

    # Получение данных и отрисовка графика
    result = get_plots(physics_path, math_path, stat_path, model_name, ax)
    if result:
        datas.append(result)
    else:
        # Если данных нет, показываем пустой график
        ax.text(0.5, 0.5, 0.5, 'No Data', transform=ax.transAxes,
               ha='center', va='center', fontsize=12)
        ax.set_xlabel('PC1')
        ax.set_ylabel('PC2')
        ax.set_zlabel('PC3')
        ax.set_title(f'{model_name}\nLast Layer', fontsize=12)

# Общий заголовок для всей фигуры
fig.suptitle('3D PCA: Physics vs Math vs Statistics Across Models\n(Last Layer Representations)',
             fontsize=16, y=0.98)

# Регулировка расположения подграфиков
plt.tight_layout(rect=[0, 0, 1, 0.96])  # Оставляем место для общего заголовка
plt.subplots_adjust(hspace=0.3, wspace=0.2)  # Увеличил отступы между графиками

plt.show()
```
